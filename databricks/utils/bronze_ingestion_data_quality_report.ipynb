```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

def bronze_data_quality_report(data_source: str, unique_id_column: str, non_null_columns: list[str]):
    print(f"🥉 BRONZE LAYER DATA QUALITY REPORT FOR: {data_source}")
    print("=" * 50)

    df = spark.table(data_source)
    print(f"Total records: {df.count():,}")
    print(f"   Schema columns: {len(df.columns)}")

    # check for duplicates
    unique_ids = df.select(unique_id_column).distinct().count()
    print(f"   Unique IDs: {unique_ids:,}")
    print(f"   Duplicates: {df.count() - unique_ids:,}")

    for column in non_null_columns:
        null_count = df.filter(col(column).isNull()).count()
        print(f"   Null {column}: {null_count:,}")

    print(f"Bronze layer data quality report complete for {data_source}.")



```
